# Utiliser le modèle intégré

Effectuer des prédictions avec les deux artefacts fournis dans `models/`, à partir de données brutes.


```python
from pathlib import Path
import sys

ROOT = Path.cwd() if (Path.cwd() / 'app.py').exists() else Path.cwd().parent
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

import pandas as pd
from app import validate_data
from app import predict_customers

customers = pd.read_csv(ROOT / "static/examples/customers.csv")
customers
```

```python
features = validate_data(customers)
result = features.copy()
result["Churn Prediction"] = ["Churn" if value == 1 else "Not Churn" for value in predict_customers(features)]
result
```

Les sorties sont des classes prédites, pas des probabilités calibrées. L’exemple est fictif et ne constitue pas un jeu d’évaluation.
