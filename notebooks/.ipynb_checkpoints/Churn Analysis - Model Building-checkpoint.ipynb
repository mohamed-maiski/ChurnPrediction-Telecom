```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.feature_selection import SelectKBest, chi2
from sklearn.metrics import *
from sklearn.tree import DecisionTreeClassifier, export_graphviz
from sklearn.svm import SVC
from sklearn.linear_model import LogisticRegression


# If you need to define a dataset or perform any further operations, you can do so below this line.
```

```python
df =pd.read_csv('tel_churn.csv')
df.head()
```

```python
X = df.drop(columns = "Churn")
y = df["Churn"]
X.head()
```

```python
SelecteFeature = SelectKBest(chi2 ,k=14)
X=SelecteFeature.fit_transform(X, y)
```

```python
X.shape
```

```python
F=SelecteFeature.get_support()
F
```

```python
df=list(df.columns)
```

```python
from imbalanced_dataset_sampler import SMOTETomek

# Créer une instance de SMOTETomek
smotetomek = SMOTETomek()

# Appliquer SMOTETomek pour sur-échantillonner et sous-échantillonner les données
X_resampled, y_resampled = smotetomek.fit_resample(X, y)
```

```python
j=0
for i in F:
    if i == True:
        print(df[j])
    j+=1
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.2)
```

```python
model_rf_smote=RandomForestClassifier(n_estimators=100, criterion='gini', random_state = 100,max_depth=6, min_samples_leaf=8)
```

```python
model_rf_smote.fit(x_train,y_train)
```

```python
yr_predict1 = model_rf_smote.predict(x_test)
```

```python
model_score_r1 = model_rf_smote.score(x_test, y_test)
```

```python
print(model_score_r1)
print(metrics.classification_report(y_test, yr_predict1))
```

```python
filename = 'model.pkl'
import pickle
```

```python
pickle.dump(model_rf_smote, open(filename, 'wb'))
```
